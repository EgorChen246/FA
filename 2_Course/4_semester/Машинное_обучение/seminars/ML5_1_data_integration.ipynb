```python
df_combined.plot(
    kind='density',
    subplots=True,
    layout=(4, 4),   # 4 строки по 4 столбца = 16 графиков
    sharex=False,    # Разные шкалы для каждого графика
    figsize=(16, 16) # Увеличиваем размер графика
)

plt.subplots_adjust(hspace=0.5, wspace=0.3) # Отступы между графиками
plt.show()
```

```python
df_numeric = df_combined.select_dtypes(include=np.number)
```

```python
fig, axes = plt.subplots(4, 4, figsize=(16, 16))
axes = axes.flatten()

num_numeric_cols = len(df_numeric.columns)

if num_numeric_cols > len(axes):
    print(f"Warning: Not enough subplots for all {num_numeric_cols} numeric columns. Plotting first {len(axes)}.")
    plot_cols = df_numeric.columns[:len(axes)]
else:
    plot_cols = df_numeric.columns


for i, col in enumerate(plot_cols):
    df_numeric[col].plot(kind='density', ax=axes[i])
    axes[i].set_title(col)
    axes[i].grid(True, alpha=0.3)


for j in range(len(plot_cols), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

#### 5.Изучите документацию pandas в части описания методов merge и join.

pandas.merge(left, right, how='inner', on=None, left_on=None, right_on=None, left_index=False, right_index=False, sort=False, suffixes=('_x', '_y'), copy=None, indicator=False, validate=None)

Parameters:
leftDataFrame or named Series
rightDataFrame or named Series
Object to merge with.

how{‘left’, ‘right’, ‘outer’, ‘inner’, ‘cross’}, default ‘inner’
Type of merge to be performed.

left: use only keys from left frame, similar to a SQL left outer join; preserve key order.

right: use only keys from right frame, similar to a SQL right outer join; preserve key order.

outer: use union of keys from both frames, similar to a SQL full outer join; sort keys lexicographically.

inner: use intersection of keys from both frames, similar to a SQL inner join; preserve the order of the left keys.

cross: creates the cartesian product from both frames, preserves the order of the left keys. <br>

Merge - Объединение по значениям столбцов, значения в таблицах должны быть одного типа данные. <br>
