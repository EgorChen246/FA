```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
```

```python
data = load_breast_cancer()
```

```python
print(data.DESCR)
```

Датасет содержит 569 образцов, 30 числовых признаков (характеристики клеток опухоли).  

Целевая переменная — бинарна классификацияя: 0 = злокачественная (malignant), 1 = доброкачественная (benign)

```python
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

Пропусков нет, все признаки числовые

```python
print(df['target'].value_counts(normalize=True))
```

```python
X = data.data
y = data.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
```

```python
print(f'Размер обучающей выборки: {X_train.shape[0]}')
print(f'Размер тестовой выборки: {X_test.shape[0]}')
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

### Логистической регрессия

```python
model = LogisticRegression(max_iter=1000, random_state=42)
```

```python
model.fit(X_train_scaled, y_train)
```

```python
y_pred = model.predict(X_test_scaled)
y_pred_proba = model.predict_proba(X_test_scaled)[:, 1]
```

### Вывод нескольких первых теоретических и эмпирических значений

```python
comparison_df = pd.DataFrame({
    'Истинное значение (y_test)': y_test[:10],
    'Предсказанное значение (y_pred)': y_pred[:10],
    'Вероятность класса 1': y_pred_proba[:10]
})
```

```python
comparison_df
```

Первые 10 предсказаний модели полностью совпадают с истинными значениями. Это указывает на высокое качество модели на тестовой выборке

```python
print(f'Accuracy: {accuracy_score(y_test, y_pred):.4f}')
print(f'Precision: {precision_score(y_test, y_pred):.4f}')
print(f'Recall: {recall_score(y_test, y_pred):.4f}')
print(f'F1-score: {f1_score(y_test, y_pred):.4f}')
```

Модель логистической регрессии показала отличные результаты на тестовой выборке:

Accuracy (точность): ~98% — модель правильно классифицирует 98 из 100 случаев.  
Precision (точность предсказания): ~99% — когда модель предсказывает "доброкачественную", она ошибается только в 2% случаев.  
Recall (полнота): ~99% — модель находит 99% всех доброкачественных опухолей.  
F1-score: ~99% — гармоническое среднее между Precision и Recall.

```python
print(classification_report(y_test, y_pred, target_names=data.target_names))
```

```python
sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt='d', cmap='Blues')
plt.xlabel('Предсказанные')
plt.ylabel('Действительные')
plt.title('Матрица ошибок')
plt.show()
```

Матрица ошибок:

True Negatives (TN): 64 — правильно предсказаны злокачественные.  
False Positives (FP): 1 — только 1 случай, когда модель ошиблась.  
True Positives (TP): 107 — правильно предсказаны доброкачественные.  
False Negatives (FN): 1 — только 1 пропущенный случай.

Модель логистической ресгресси отлично подошла по наш датасет. Минимальное кол-во ошибок 2 из 171
