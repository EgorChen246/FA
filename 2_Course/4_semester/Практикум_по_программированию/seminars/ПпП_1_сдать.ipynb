```python
class Software:  # класс программное_обеспечение
    def __init__(self, name, vendor):
        self.name = name
        self.vendor = vendor

    def show(self):
        print(f"PO: {self.name}, Vendor: {self.vendor}")

    def check(self, today):
        return True

class Free(Software):  # дочерний класс свободное
    def show(self):  # название, производитель
        print(f"Свободное: {self.name} | {self.vendor}")

class Shareware(Software):  # дочерний класс условно-бесплатная
    def __init__(self, name, vendor, install_day, limit):
        super().__init__(name, vendor)
        self.install_day = install_day
        self.limit = limit

    def show(self):  # название, производитель, дата установки, срок бесплатного исп
        print(f"Условно-бесплатная: {self.name} | {self.vendor} | дата установки: {self.install_day} | Срок: {self.limit}")

    def check(self, today):
        return today <= (self.install_day + self.limit)

class Commercial(Software):  # дочерний класс коммерческое
    def __init__(self, name, vendor, price, install_day, limit):
        super().__init__(name, vendor)
        self.price = price
        self.install_day = install_day
        self.limit = limit

    def show(self):  # название, производитель, цена, дата установки, срок использования
        print(f"Коммерческое: {self.name} | {self.vendor} | Цена: {self.price} | дата установки: {self.install_day} | Срок: {self.limit}")

    def check(self, today):
        return today <= (self.install_day + self.limit)

progs = [
    Free("Linux", "UNIX"),
    Shareware("WinZip", "Corel", 10, 30),
    Commercial("MS Office", "Microsoft", 15000, 5, 365),
    Shareware("TotalCmd", "Ghisler", 1, 15),
    Commercial("Games", "Steam", 2000, 50, 10)
]

now = 25 # Ввести день года

print("Список всей базы:\n")
for p in progs:
    p.show()

print("\nМожно исп сегодня:\n")
for p in progs:
    if p.check(now):
        p.show()
```

```python
import math
```

```python
def get_dist(p):
    return math.sqrt(p[0]**2 + p[1]**2)
```

```python
def bubble_sort(arr, rev):  # Сортировка пузырьком
    n = len(arr)
    for i in range(n):
        for j in range(0, n-i-1):
            if (get_dist(arr[j]) > get_dist(arr[j+1])) != rev:  # Сравниваем расстояния
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr
```

```python
def selection_sort(arr, rev):  # Сортировка выбором
    for i in range(len(arr)):
        idx = i
        for j in range(i + 1, len(arr)):
            if (get_dist(arr[j]) < get_dist(arr[idx])) != rev:
                idx = j
        arr[i], arr[idx] = arr[idx], arr[i]
    return arr
```

```python
def quick_sort(arr, rev):  # Быстрая сортировка 
    if len(arr) <= 1:
        return arr
    p = arr[len(arr) // 2]
    d_p = get_dist(p)
    
    left = [x for x in arr if (get_dist(x) < d_p) != rev]
    middle = [x for x in arr if get_dist(x) == d_p]
    right = [x for x in arr if (get_dist(x) > d_p) != rev]
```

Открытие и создание файла (points.txt)

```python
points = []
try:
    with open("points.txt", "r") as f:
        for line in f:
            points.append(list(map(float, line.split())))
except FileNotFoundError:
    print("Файл не найден")
    points = [[3, 4], [1, 1], [0, 5], [2, 2]]
```

```python
order = input("Сортировать по возрастанию (1) или убыванию (2) ") == "2"
algo = input("Алгоритм: 1-Пузырек, 2-Выбор, 3-Быстрая: ")

if algo == "1":
    res = bubble_sort(points[:], order)
elif algo == "2":
    res = selection_sort(points[:], order)
else:
    res = quick_sort(points[:], order)

print("\nРезультат (X, Y | Расстояние): \n")
for p in res:
    print(f"{p[0]}, {p[1]} | Расстояние: {get_dist(p):.2f}")
```

```python
def is_string(val):  # условие (проверка на строку)
    return isinstance(val, str)

def to_upper(val):  # преобразователь (в верхний регистр)
    return val.upper()
```

Декаратор

```python
def modify_args(check_func, transform_func):
    def decorator(original_func):
        def wrapper(arg):
            
            if check_func(arg):  #  если верно, то меняем
                arg = transform_func(arg)
                
            return original_func(arg)  
        return wrapper
    return decorator
```

Исп декаратор

```python
@modify_args(is_string, to_upper)
def print_info(data):
    print(f"Результат функции: \n {data}")

# Проверка
print("Тест со строкой")
print_info("Да, я на чилле, на расслабоне \n \
Я не спешу, времени море \n \
Делаю то, что хочу, ты понял \n \
Двести пропущенных на телефоне \n \
Белый песок под моими ногами \n \
Жаркое лето, мы отдыхаем")  


print("\nТест с числом")
print_info(123)
```
